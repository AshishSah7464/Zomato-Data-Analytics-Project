```python
import pandas as pd
import numpy as np
```

```python
data=pd.read_csv('../data/raw/Zomato.csv')
print("Dataset loaded successfully!")
```

```python
df=data.copy()
print('Data Copy Successfully')
```

```python
df.shape
```

```python
df.head()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates(keep='first')
print("Duplicate Rows Deleted Succesfully")
```

```python
# verify Duplicated
df.duplicated().sum()
```

```python
df.shape
```

```python
df['rate']
```

```python
# Replace '/5' with empty string
df["rate"] = df["rate"].str.replace("/5", "", regex=False)
```

```python
#convert to numeric
df["rate"] = pd.to_numeric(df["rate"], errors="coerce")
```

```python
df['rate']
```

```python
print(df['rate'].dtype)
```

```python
# check unique values for categorical columns
print(df["online_order"].unique())
print(df["book_table"].unique())
print(df["listed_in(type)"].unique())
```

```python
df["rate"].describe()
```

```python
# check the rate is above 5
df[df["rate"] > 5]
```

```python
# check the rate is below 5
df[df["rate"] < 0]
```

```python
df.describe()
```

```python
df = df.rename(columns={
    "approx_cost(for two people)": "approx_cost",
    "listed_in(type)": "restaurant_type"
})
```
